<div class="skillos-primary-heading">Introduction</div>

<div class="skillos-primary-text">
SkillOS is an experimental operating system for AI agents—not in the traditional sense, but as an orchestration framework that provides operating-system-like services for planning, skill management, and runtime execution. It converts business requests into structured, reusable, and executable workflows by organizing AI capabilities as skills that can be planned, generated, reviewed, registered, and reused across future tasks.
</div>

<div class="skillos-primary-text">
The central idea is simple: when a user provides a request, SkillOS first creates an execution plan. If a required capability already exists, the system routes the task to the registered skill. If the capability does not exist, the Skill Factory can generate a new SKILL.md draft, evaluate it with an LLM-as-Judge, and require human approval before it becomes part of the skill registry.
</div>

<div style="text-align:center; margin:30px 0;">

</div>

<div class="skillos-primary-text">
The architecture separates planning, skill creation, validation, registry management, runtime execution, and external tool access. This separation keeps the system modular: skills define what should be done, while runtime modules perform operational tasks such as project initialization, webpage generation, image generation, report writing, and GitHub publishing.
</div>

<div class="skillos-primary-text">
  This workflow shows how SkillOS turns general knowledge into executable action. A reusable <span class="skillos-highlight-primary">Master Skill</span> defines the overall capability, reasoning pattern, and quality standards. For a specific user request, SkillOS derives a tailored <span class="skillos-highlight-primary">Task Skill</span> from that master template, then executes the Task Skill to produce the final result. Once approved, the generated skill can be registered and reused in future workflows.
</div>

<div class="skillos-primary-text">
Several core components work together in the notebook: the Planning Module creates ordered workflow steps; the Skill Factory generates new skill definitions; the Skill Registry stores approved capabilities; the Human Approval Gate prevents automatic registration; the LLM-as-Judge evaluates generated artifacts; and runtime modules execute structured commands. GitHub integration is handled through MCP, allowing repository creation, file updates, and website publishing.

````python
(RUNTIME_DIR / "image_renderer.py").write_text(
'''
from typing import Any
from pathlib import Path

import torch
from diffusers import FluxPipeline
from IPython.display import display


def generate_images(
    command: dict[str, Any],
    hf_token: str,
) -> list[str]:
    """
    Generate images from a structured runtime command.

    Description
    -----------
    Uses the FLUX image generation model to create one or more images
    from the supplied prompt, saves them to disk, displays each image
    in the notebook, and returns the saved file paths.

    Args
    ----
    command:
        Dictionary containing the image generation parameters.

        Expected fields include:

        - prompt
        - num_images
        - width
        - height
        - image_format
        - output_folder
        - filename_prefix
        - seed

    hf_token:
        Hugging Face access token used to download the FLUX model.

    Returns
    -------
    list[str]

        Paths of the generated image files.

    Raises
    ------
    RuntimeError

        If image generation fails.

    Example
    -------
    >>> generate_images(
    ...     command,
    ...     hf_token,
    ... )
    """

    # -----------------------------------------------------
    # Read command parameters
    # -----------------------------------------------------

    prompt = command["prompt"]
    num_images = command.get("num_images", 1)
    width = command.get("width", 512)
    height = command.get("height", 512)
    image_format = command.get("image_format", "png")
    output_folder = Path(
        command.get(
            "output_folder",
            "/kaggle/working/images",
        )
    )
    filename_prefix = command.get(
        "filename_prefix",
        "image",
    )
    seed = command.get("seed", 42)

    # -----------------------------------------------------
    # Create output directory
    # -----------------------------------------------------

    output_folder.mkdir(
        parents=True,
        exist_ok=True,
    )

    # -----------------------------------------------------
    # Select device and tensor precision
    # -----------------------------------------------------

    device = (
        "cuda"
        if torch.cuda.is_available()
        else "cpu"
    )

    dtype = (
        torch.bfloat16
        if device == "cuda"
        else torch.float32
    )

    # -----------------------------------------------------
    # Load the FLUX image generation pipeline
    # -----------------------------------------------------

    pipe = FluxPipeline.from_pretrained(
        "black-forest-labs/FLUX.1-schnell",
        torch_dtype=dtype,
        token=hf_token,
    ).to(device)

    # pipe.enable_model_cpu_offload()
    pipe.enable_attention_slicing()
    pipe.enable_vae_slicing()
    pipe.enable_vae_tiling()

    # -----------------------------------------------------
    # Create reproducible random generator (optional)
    # -----------------------------------------------------

    generator = None

    if seed is not None:
        generator = (
            torch.Generator(device=device)
            .manual_seed(seed)
        )

    # -----------------------------------------------------
    # Generate images
    # -----------------------------------------------------

    saved_files: list[str] = []

    for index in range(num_images):

        image = pipe(
            prompt=prompt,
            width=width,
            height=height,
            guidance_scale=0.0,
            num_inference_steps=4,
            generator=generator,
        ).images[0]

        file_path = output_folder / (
            f"{filename_prefix}_{index + 1}.{image_format}"
        )

        image.save(file_path)

        display(image)

        saved_files.append(
            str(file_path)
        )

    # -----------------------------------------------------
    # Return generated file paths
    # -----------------------------------------------------

    return saved_files
'''
)
````

### Runtime dispatcher

````python
def execute_runtime_command_if_needed(
    output: Any,
) -> Any:
    """
    Execute a runtime command produced by a skill, if present.

    Description
    -----------
    Checks whether the supplied output is a runtime command generated
    by a SkillOS skill. If the output represents a
    ``run_python_script`` command, the corresponding runtime function
    is looked up in the runtime registry and executed. Otherwise, the
    original output is returned unchanged.

    Args
    ----
    output:
        Output produced by a SkillOS skill. This may be a Pydantic
        model representing a runtime command or any other object.

    Returns
    -------
    Any

        The result of the executed runtime command if one was found;
        otherwise, the original output.

    Raises
    ------
    RuntimeError

        If the runtime command does not specify a script or if the
        specified runtime script is not registered.

    Example
    -------
    >>> execute_runtime_command_if_needed(output)
    '/kaggle/working/project-catflix'
    """

  
    # -----------------------------------------------------
    # Ignore outputs that are not runtime command schemas
    # -----------------------------------------------------

    if not isinstance(
        output,
        BaseModel,
    ):
        return output

    command = output.model_dump()

    # -----------------------------------------
    # Cache the current project name
    # -----------------------------------------

    global PROJECT

    if command.get("script") == "runtime/project_initializer.py":

        PROJECT = command["args"]["project_name"]

    # -----------------------------------------------------
    # Ignore non-runtime commands
    # -----------------------------------------------------

    if command.get("name") != "run_python_script":
        return output

    # -----------------------------------------------------
    # Retrieve the runtime script
    # -----------------------------------------------------

    script = command.get("script")

    if script is None:
        raise RuntimeError(
            "Runtime command is missing the 'script' field."
        )

    # -----------------------------------------------------
    # Look up the runtime function
    # -----------------------------------------------------

    runtime_function = RUNTIME_REGISTRY.get(script)

    if runtime_function is None:
        raise RuntimeError(
            f"Unknown runtime script: {script}"
        )

    # -----------------------------------------------------
    # Execute the runtime command
    # -----------------------------------------------------

    return runtime_function(command)
````

### Executing a single ADK agent

````python
async def run_agent_once(
    agent: Agent,
    prompt: str,
    step_num: int,
    *,
    agent_name: str,
) -> str:
    """
    Execute a single ADK agent and return its final response.

    Description
    -----------
    Creates an isolated ADK runtime session, executes the supplied
    agent against the given prompt, collects the final response,
    and returns it as a single string.

    A fresh session is created for every invocation to ensure that
    agent executions remain independent from one another.

    Args
    ----
    agent:
        ADK agent to execute.

    prompt:
        User prompt supplied to the agent.

    step_num:
        Sequential workflow step number.

    agent_name:
        Logical runtime name used when constructing the temporary
        application name.

    Returns
    -------
    str
        Final textual response produced by the agent.

    Raises
    ------
    RuntimeError
        If the agent finishes without producing a final response.

    Example
    -------
    >>> response = await run_agent_once(
    ...     agent=task_executor_agent,
    ...     prompt="Create a landing page.",
    ...     step_num=1,
    ...     agent_name="task_executor",
    ... )

    Notes
    -----
    This function performs no output parsing or validation.
    It simply executes the agent and returns its textual output.
    """

    # -----------------------------------------------------
    # Create isolated runtime session
    # -----------------------------------------------------

    app_name = f"{agent_name}_app_{step_num}"
    user_id = "user_1"
    session_id = str(uuid.uuid4())

    session_service = InMemorySessionService()

    # -----------------------------------------------------
    # Initialize runner
    # -----------------------------------------------------

    runner = Runner(
        agent=agent,
        app_name=app_name,
        session_service=session_service,
    )

    await session_service.create_session(
        app_name=app_name,
        user_id=user_id,
        session_id=session_id,
    )

    # -----------------------------------------------------
    # Build runtime message
    # -----------------------------------------------------

    message = types.Content(
        role="user",
        parts=[
            types.Part(
                text=prompt,
            )
        ],
    )

    # -----------------------------------------------------
    # Execute agent
    # -----------------------------------------------------

    response_chunks: list[str] = []

    # Pause to respect API rate limits before the LLM call.
    await asyncio.sleep(API_REQUEST_DELAY)

    async for event in runner.run_async(
        user_id=user_id,
        session_id=session_id,
        new_message=message,
    ):

        if (
            event.is_final_response()
            and event.content
            and event.content.parts
        ):

            segment = "\n".join(
                part.text
                for part in event.content.parts
                if part.text
            )

            if segment:
                response_chunks.append(segment)
                
    print("▶️ Execution step - 5")
    
    # -----------------------------------------------------
    # Validate response
    # -----------------------------------------------------

    if not response_chunks:
        raise RuntimeError(
            "The agent completed without producing a final response."
        )

    # -----------------------------------------------------
    # Return response
    # -----------------------------------------------------

    return "\n\n".join(response_chunks)
````

### Executing a single OpenAI agent

````python
async def run_openai_skill_manager(
    skill_path: Path,
    prompt: str,
    model: str = "gpt-5",
) -> str:
    """
    Execute a SkillOS skill using an OpenAI model.

    Description
    -----------
    Loads the specified SKILL.md file, constructs the Skill Manager
    prompt, submits it to the OpenAI Responses API, and returns the
    generated response.

    This function is intended for SkillOS skills that are executed
    directly by an OpenAI model rather than through the Google ADK
    runtime.

    Args
    ----
    skill_path:
        Path to the skill directory containing the SKILL.md file.

    prompt:
        User request to be executed by the skill.

    model:
        OpenAI model used to execute the skill.
        Defaults to "gpt-5".

    Returns
    -------
    str

        Final textual response generated by the Skill Manager.

    Raises
    ------
    FileNotFoundError

        If the supplied skill directory or SKILL.md file does not
        exist.

    Example
    -------
    >>> response = await run_openai_skill_manager(
    ...     skill_path=Path("./skills/web-file-generator"),
    ...     prompt="Generate a landing page.",
    ... )
    """

    # -----------------------------------------------------
    # Validate the skill directory
    # -----------------------------------------------------

    skill_path = Path(skill_path)

    if not skill_path.exists():
        raise FileNotFoundError(
            f"Skill directory not found: {skill_path}"
        )

    skill_md_path = skill_path / "SKILL.md"

    if not skill_md_path.exists():
        raise FileNotFoundError(
            f"SKILL.md not found: {skill_md_path}"
        )

    # -----------------------------------------------------
    # Load the skill definition
    # -----------------------------------------------------

    skill_md = skill_md_path.read_text(
        encoding="utf-8",
    )

    # -----------------------------------------------------
    # Build the Skill Manager prompt
    # -----------------------------------------------------

    skill_manager_prompt = build_skill_manager_prompt(
        skill_md=skill_md,
        prompt=prompt,
    )

    # -----------------------------------------------------
    # Execute the skill using the OpenAI Responses API
    # -----------------------------------------------------

    response = openai_client.responses.create(
        model=model,
        input=skill_manager_prompt,
    )

    # -----------------------------------------------------
    # Return the generated response
    # -----------------------------------------------------

    return response.output_text
````

### Running an agent with the skill given the user prompt 

````python
async def run_skill_agent(
    skill_path: Path,
    prompt: str,
    step_num: int,
    output_schema: type[BaseModel] | None = None,
) -> str | BaseModel:  
    """
    Execute a single SkillOS skill.

    Description
    -----------
    Loads the requested skill from disk, creates an isolated ADK
    execution environment, executes the skill, and optionally
    reformats the result into a structured JSON object matching
    the supplied output schema.

    Execution Pipeline
    ------------------
        User Prompt
              │
              ▼
        Skill Manager Agent
              │
              ▼
         Raw Text Output
              │
              ▼
      (Optional Formatter Agent)
              │
              ▼
      Structured JSON Output

    Args
    ----
    skill_path:
        Path to the skill directory containing SKILL.md.

    prompt:
        Prompt sent to the skill.

    step_num:
        Sequential workflow step number.

    output_schema:
        Optional Pydantic schema describing the expected structured
        output. If omitted, the raw textual response is returned.

    Returns
    -------
    str | BaseModel
        Raw text when no output schema is supplied, otherwise an
        instance of the supplied Pydantic model.

    Raises
    ------
    FileNotFoundError
        If the supplied skill directory does not exist.

    RuntimeError
        If the runtime fails to generate a final response.

    Example
    -------
    >>> result = await run_skill_agent(
    ...     skill_path=Path("./skills/project-initializer"),
    ...     prompt="Create a new project.",
    ...     step_num=1,
    ... )
    """
    print("▶️ Execution step - 3")
    
    # -----------------------------------------------------
    # Validate skill directory
    # -----------------------------------------------------

    if not skill_path.exists():
        raise FileNotFoundError(
            f"Skill directory not found: {skill_path}"
        )

    
    # =====================================================
    # ALTERNATIVE 1:
    # =====================================================
    
    # -----------------------------------------------------
    # Load runtime skill
    # -----------------------------------------------------

    runtime_skill = load_skill_from_dir(
        str(skill_path)
    )

    runtime_skill_toolset = SkillToolset(
        skills=[runtime_skill]
    )

    # -----------------------------------------------------
    # Create execution model
    # -----------------------------------------------------

    skill_manager_model = Gemini(
            model=SKILL_MANAGER_MODEL, 
            api_key=os.environ["GOOGLE_API_KEY"],
    )

    # -----------------------------------------------------
    # Skill execution agent
    # -----------------------------------------------------

    skill_manager_agent = Agent(
        name="skill_manager",
        model=skill_manager_model,
        description=(
            "Executes the loaded SkillOS skill."
        ),
        instruction=(
            "Use only the loaded skill.\n"
            "Follow the skill instructions exactly.\n"
            "Return only the final result required by the skill.\n"
        ),
        tools=[runtime_skill_toolset],
    )

    # -----------------------------------------------------
    # Execute the skill
    # -----------------------------------------------------

    skill_manager_response = await run_agent_once(
            agent=skill_manager_agent,
            prompt=prompt,
            step_num=step_num,
            agent_name="skill_manager",
    )


    # =====================================================
    # ALTERNATIVE 2:
    # =====================================================
    
    # skill_manager_response = await run_openai_skill_manager(
    #     skill_path=skill_path,
    #     prompt=prompt,
    #     model=SKILL_MANAGER_MODEL,
    # )   
    
    # -----------------------------------------------------
    # Return raw response
    # -----------------------------------------------------

    if output_schema is None:
        return skill_manager_response

    # -----------------------------------------------------
    # Formatter agent
    # -----------------------------------------------------

    json_formatter_model = Gemini(
        model=FORMATTER_MODEL,
        api_key=os.environ["GOOGLE_API_KEY"],
    )

    json_formatter_agent = Agent(
        name="json_formatter",
        model=json_formatter_model,
        description=(
            "Formats text into valid JSON."
        ),
        instruction=(
            "Convert the supplied text into JSON that "
            "matches the required output schema exactly.\n"
            "Return ONLY the JSON object.\n"
            "Do not include explanations, markdown, or "
            "additional text."
        ),
    )

    formatter_prompt = f"""
        Convert the following text into a JSON object that EXACTLY matches the schema below.
        Return ONLY the JSON object.
        Schema:
        {output_schema.model_json_schema()}
        Text:
        {skill_manager_response}
    """

    # -----------------------------------------------------
    # Respect API rate limits
    # -----------------------------------------------------

    await asyncio.sleep(API_REQUEST_DELAY)

    # -----------------------------------------------------
    # Format the response
    # -----------------------------------------------------
    
    print("▶️ Execution step - 4")
    
    formatted_response = await run_agent_once(
        agent=json_formatter_agent,
        prompt=formatter_prompt,
        step_num=step_num,
        agent_name="json_formatter",
    )
    
    print("▶️ Execution step - 6")
    
    # -----------------------------------------------------
    # Validate structured output
    # -----------------------------------------------------

    json_text = extract_json(formatted_response)
    
    return output_schema.model_validate(json_text)
````

### Generating a valid and unique skill ID from the skill name

````python
def sanitize_skill_id(text: str) -> str:
    """
    Convert an arbitrary skill name into a filesystem-safe kebab-case ID.

    Description
    -----------
    This function converts a raw skill name or identifier into a stable
    lowercase kebab-case value suitable for folder names, registry IDs,
    and skill paths.

    Args
    ----
    text:
        Raw skill name or identifier.

    Returns
    -------
    str
        Sanitized kebab-case skill identifier.

    Raises
    ------
    ValueError
        If the input is empty or cannot produce a valid skill ID.

    Example
    -------
    >>> sanitize_skill_id("Market Report Generator")
    'market-report-generator'

    Notes
    -----
    This function is important because SkillOS uses the same ID for:
    - registry id
    - folder name
    - skill path
    """
    if not text or not text.strip():
        raise ValueError("Skill ID text cannot be empty.")

    sanitized = text.strip().lower()
    sanitized = re.sub(r"[^a-z0-9]+", "-", sanitized)
    sanitized = re.sub(r"-+", "-", sanitized)
    sanitized = sanitized.strip("-")

    if not sanitized:
        raise ValueError(f"Could not sanitize skill ID from input: {text!r}")

    return sanitized
````

### Extracting a Python object from a JSON output

````python
def extract_json(text: str) -> Any:
    """
    Extract and parse the first valid JSON object or array from an LLM response.

    Supports:
    - Raw JSON.
    - Markdown-fenced JSON.
    - JSON surrounded by explanatory text.
    - Responses containing extra text before or after the JSON.
    """

    if not text or not text.strip():
        raise ValueError("Cannot extract JSON from empty text.")

    cleaned = text.strip()

    # -----------------------------------------------------
    # 1. Try parsing the entire response.
    # -----------------------------------------------------

    try:
        return json.loads(cleaned)
    except json.JSONDecodeError:
        pass

    # -----------------------------------------------------
    # 2. Try JSON inside Markdown code fences.
    # -----------------------------------------------------

    fence_match = re.search(
        r"```(?:json)?\s*(.*?)\s*```",
        cleaned,
        re.DOTALL | re.IGNORECASE,
    )

    if fence_match:
        try:
            return json.loads(fence_match.group(1).strip())
        except json.JSONDecodeError:
            pass

    # -----------------------------------------------------
    # 3. Try everything between the first '{' and last '}'
    #    (or '[' and ']').
    # -----------------------------------------------------

    first_object = cleaned.find("{")
    last_object = cleaned.rfind("}")

    if first_object != -1 and last_object != -1:
        try:
            return json.loads(cleaned[first_object:last_object + 1])
        except json.JSONDecodeError:
            pass

    first_array = cleaned.find("[")
    last_array = cleaned.rfind("]")

    if first_array != -1 and last_array != -1:
        try:
            return json.loads(cleaned[first_array:last_array + 1])
        except json.JSONDecodeError:
            pass

    # -----------------------------------------------------
    # 4. Fall back to JSONDecoder.
    # -----------------------------------------------------

    decoder = json.JSONDecoder()

    for index, char in enumerate(cleaned):
        if char in "{[":
            try:
                obj, _ = decoder.raw_decode(cleaned[index:])
                return obj
            except json.JSONDecodeError:
                continue

    raise ValueError(
        "No valid JSON object or array found in model output."
    )
````

### Extracting the metadata from the generated SKILL.md

````python
def extract_yaml_frontmatter(markdown_text: str) -> Dict[str, Any]:
    """
    Extract YAML frontmatter from a generated SKILL.md document.

    Description
    -----------
    Generated skills are expected to begin with YAML frontmatter
    enclosed between two `---` lines. This metadata is used as the
    source of truth for the registry entry.

    Args
    ----
    markdown_text:
        Complete SKILL.md text generated by the Skill Factory.

    Returns
    -------
    Dict[str, Any]
        Parsed YAML metadata dictionary.

    Raises
    ------
    ValueError
        If no frontmatter exists or if the frontmatter is not a valid
        YAML dictionary.

    Example
    -------
    >>> text = "---\\nid: report-writer\\nname: Report Writer\\n---\\n# Skill"
    >>> extract_yaml_frontmatter(text)["id"]
    'report-writer'

    Notes
    -----
    This function assumes the frontmatter appears at the very beginning
    of the file.
    """
    if not markdown_text or not markdown_text.strip():
        raise ValueError("Cannot extract YAML frontmatter from empty markdown.")
   
    match = FRONTMATTER_PATTERN.search(markdown_text.strip())

    if not match:
        raise ValueError("No YAML frontmatter found in generated SKILL.md.")

    metadata_text = match.group(1)
    metadata = yaml.safe_load(metadata_text)

    if not isinstance(metadata, dict):
        raise ValueError("YAML frontmatter is not a valid dictionary.")

    return metadata
````

### Getting web files from the directory

````python
def get_web_files(
    project_dir: Path,
) -> list[dict[str, Any]]:
    """
    Read all files under the project's web directory.

    Args
    ----
    project_dir:
        Root directory of the project.

    Returns
    -------
    list[dict[str, Any]]

        List of file objects suitable for the GitHub MCP
        push_files tool.

        Example:

        [
            {
                "path": "index.html",
                "content": "<!DOCTYPE html>..."
            },
            {
                "path": "css/styles.css",
                "content": "..."
            }
        ]
    """

    web_dir = project_dir / "web"

    if not web_dir.exists():
        raise FileNotFoundError(
            f"Web directory does not exist: {web_dir}"
        )

    files: list[dict[str, Any]] = []

    for file_path in web_dir.rglob("*"):

        if not file_path.is_file():
            continue

        files.append(
            {
                "path": file_path.relative_to(web_dir).as_posix(),
                "content": file_path.read_text(
                    encoding="utf-8",
                ),
            }
        )

    return files
````

### Loading existing skills from the registry file

````python
def load_registry(registry_path: Path) -> List[Dict[str, Any]]:
    """
    Load the SkillOS skills registry from disk.

    Description
    -----------
    Reads `skills_registry.yaml` and returns the list stored under the
    top-level `skills` key. If the registry does not exist yet, this
    function returns an empty list.

    Args
    ----
    registry_path:
        Path to the YAML registry file.

    Returns
    -------
    List[Dict[str, Any]]
        List of registered skill entries.

    Raises
    ------
    ValueError
        If the YAML file exists but does not contain a valid `skills` list.

    Example
    -------
    >>> skills = load_registry(Path("/kaggle/working/skills_registry.yaml"))
    >>> isinstance(skills, list)
    True

    Notes
    -----
    The registry is an index, not the authoritative skill definition.
    The generated SKILL.md remains the source of truth.
    """
    if not registry_path.exists():
        return []

    with registry_path.open("r", encoding="utf-8") as registry_file:
        registry_data = yaml.safe_load(registry_file) or {}

    skills = registry_data.get("skills", [])

    if not isinstance(skills, list):
        raise ValueError("Invalid registry format: top-level 'skills' must be a list.")

    return skills
````

### Rewriting the registry file with new skills

````python
def save_registry(registry_path: Path, skills: List[Dict[str, Any]]) -> None:
    """
    Save the SkillOS skills registry to disk.

    Description
    -----------
    Writes the provided list of skill entries into the central
    `skills_registry.yaml` file.

    Args
    ----
    registry_path:
        Path to the YAML registry file.

    skills:
        List of skill registry entries.

    Returns
    -------
    None

    Raises
    ------
    ValueError
        If `skills` is not a list.
    OSError
        If the file cannot be written.

    Example
    -------
    >>> save_registry(Path("skills_registry.yaml"), [{"id": "demo"}])

    Notes
    -----
    Existing registry entries should be loaded first and preserved.
    This function simply writes the supplied registry list.
    """
    if not isinstance(skills, list):
        raise ValueError("Registry data must be a list of skill entries.")

    registry_path.parent.mkdir(parents=True, exist_ok=True)

    with registry_path.open("w", encoding="utf-8") as registry_file:
        yaml.dump(
            {"skills": skills},
            registry_file,
            default_flow_style=False,
            sort_keys=False,
            allow_unicode=True,
        )
````

### Building metadata from the SKILL.md for registry entry

````python
def build_registry_entry(
    skill_metadata: Dict[str, Any],
    target_dir: Path,
) -> Dict[str, Any]:
    """
    Build a registry entry from generated SKILL.md metadata.

    Description
    -----------
    Converts parsed SKILL.md frontmatter into the normalized registry
    format expected by SkillOS.

    Args
    ----
    skill_metadata:
        Metadata parsed from the generated SKILL.md frontmatter.

    target_dir:
        Directory where the generated SKILL.md will be stored.

    Returns
    -------
    Dict[str, Any]
        Normalized registry entry.

    Raises
    ------
    ValueError
        If neither `id` nor `name` is available in the metadata.

    Example
    -------
    >>> metadata = {"id": "text-summary", "name": "Text Summary"}
    >>> build_registry_entry(metadata, Path("./skills/text-summary"))["id"]
    'text-summary'

    Notes
    -----
    Defaults are applied for optional fields such as `version`,
    `enabled`, and `execution_modes`.
    """
    raw_id = skill_metadata.get("id") or skill_metadata.get("name")

    if not raw_id:
        raise ValueError("Generated skill metadata must include at least 'id' or 'name'.")

    skill_id = sanitize_skill_id(str(raw_id))

    execution_modes = skill_metadata.get("execution_modes") or ["standalone"]

    if isinstance(execution_modes, str):
        execution_modes = [execution_modes]

    return {
        "id": skill_id,
        "name": skill_metadata.get("name") or skill_id,
        "description": skill_metadata.get("description") or "",
        "category": skill_metadata.get("category") or "general",
        "tags": skill_metadata.get("tags") or [],
        "skill_path": str(target_dir),
        "version": skill_metadata.get("version") or "1.0.0",
        "enabled": skill_metadata.get("enabled", True),
        "execution_modes": execution_modes,
        "input_types": skill_metadata.get("input_types") or [],
        "output_types": skill_metadata.get("output_types") or [],
    }
````

### Validating a new skill registry entry before append

````python
def validate_new_skill_entry(
    new_entry: Dict[str, Any],
    available_skills: List[Dict[str, Any]],
) -> None:
    """
    Validate a new skill registry entry before appending it.

    Description
    -----------
    Ensures that the new registry entry does not conflict with existing
    skills and follows the allowed execution mode rules.

    Args
    ----
    new_entry:
        Registry entry for the generated skill.

    available_skills:
        Existing registered skills.

    Returns
    -------
    None

    Raises
    ------
    ValueError
        If duplicate IDs, duplicate paths, duplicate names, or invalid
        execution modes are found.

    Example
    -------
    >>> validate_new_skill_entry({"id": "a", "name": "A", "skill_path": "./a",
    ... "execution_modes": ["standalone"]}, [])

    Notes
    -----
    Duplicate names are rejected for safety. If versioning is desired,
    explicit version-handling logic should be added later.
    """
    new_id = new_entry.get("id")
    new_name = new_entry.get("name")
    new_path = new_entry.get("skill_path")
    execution_modes = new_entry.get("execution_modes", [])

    if not new_id:
        raise ValueError("New registry entry must include an 'id'.")

    if not new_name:
        raise ValueError("New registry entry must include a 'name'.")

    if not new_path:
        raise ValueError("New registry entry must include a 'skill_path'.")

    if not isinstance(execution_modes, list):
        raise ValueError("'execution_modes' must be a list.")

    invalid_modes = set(execution_modes) - VALID_EXECUTION_MODES

    if invalid_modes:
        raise ValueError(f"Invalid execution mode(s): {sorted(invalid_modes)}")

    for skill in available_skills:
        if skill.get("id") == new_id:
            raise ValueError(f"Duplicate skill id found: {new_id}")

        if skill.get("skill_path") == new_path:
            raise ValueError(f"Duplicate skill_path found: {new_path}")

        if skill.get("name") == new_name:
            raise ValueError(f"Duplicate skill name found: {new_name}")
````

### Finding a registered skill by planner route

````python
def find_skill_by_route(
    route: str,
    available_skills: List[Dict[str, Any]],
) -> Optional[Dict[str, Any]]:
    """
    Find a registered skill by planner route.

    Description
    -----------
    The planner is instructed to use skill IDs as route values. This
    function first tries to match by `id`, then falls back to matching
    by `name` for backward compatibility.

    Args
    ----
    route:
        Route value returned by the planner.

    available_skills:
        List of registered skill entries.

    Returns
    -------
    Optional[Dict[str, Any]]
        Matching skill entry if found; otherwise `None`.

    Example
    -------
    >>> skills = [{"id": "summary", "name": "Summary"}]
    >>> find_skill_by_route("summary", skills)["name"]
    'Summary'

    Notes
    -----
    Matching by ID is preferred because names may change, while IDs
    should remain stable.
    """
    for skill in available_skills:
        if skill.get("id") == route:
            return skill

    for skill in available_skills:
        if skill.get("name") == route:
            return skill

    return None
````

### Registering a new skill

````python
def register_skill(
    full_draft: str,
    available_skills: List[Dict[str, Any]],
    skills_dir: Path = SKILLS_DIR,
    registry_path: Path = REGISTRY_FILE_PATH,
) -> Dict[str, Any]:
    """
    Register a newly generated skill.

    Description
    -----------
    This function performs the complete SkillOS registration workflow:

    1. Parse the generated SKILL.md metadata.
    2. Build the registry entry.
    3. Validate the new entry.
    4. Create the skill directory.
    5. Write the SKILL.md file.
    6. Update the central registry.
    7. Verify both filesystem and registry.

    This function assumes that:

    • the skill has already been generated;
    • the skill has already passed the LLM-as-Judge;
    • the human has already approved it.

    Args
    ----
    full_draft:
        Complete generated SKILL.md document.

    available_skills:
        Current registry loaded in memory.

    skills_dir:
        Root skills directory.

    registry_path:
        Path to the central registry.

    Returns
    -------
    Dict[str, Any]

        The newly created registry entry.

    Raises
    ------
    RuntimeError
        If verification fails.

    ValueError
        If metadata is invalid or duplicate IDs are found.

    Example
    -------
    >>> new_skill = register_skill(
    ...     full_draft=draft,
    ...     available_skills=available_skills,
    ... )
    """

    print("\n💾 Registering new skill...\n")

    # -----------------------------------------------------
    # Parse metadata
    # -----------------------------------------------------
    
    skill_metadata = extract_yaml_frontmatter(full_draft)
    
    new_skill_id = sanitize_skill_id(
        skill_metadata.get("id") or skill_metadata.get("name")
    )
    
    target_dir = skills_dir / new_skill_id
    skill_file_path = target_dir / "SKILL.md"

    # -----------------------------------------------------
    # Build registry entry
    # -----------------------------------------------------
    
    new_skill_entry = build_registry_entry(
        skill_metadata=skill_metadata,
        target_dir=target_dir,
    )
    
    # -----------------------------------------------------
    # Validate
    # -----------------------------------------------------
    
    validate_new_skill_entry(
        new_entry=new_skill_entry,
        available_skills=available_skills,
    )

    # -----------------------------------------------------
    # Create directory
    # -----------------------------------------------------
    
    target_dir.mkdir(
        parents=True,
        exist_ok=False,
    )

    # -----------------------------------------------------
    # Write SKILL.md
    # -----------------------------------------------------
    
    with skill_file_path.open(
        "w",
        encoding="utf-8",
    ) as file:

        file.write(full_draft)

    # -----------------------------------------------------
    # Update registry
    # -----------------------------------------------------

    available_skills.append(new_skill_entry)

    save_registry(
        registry_path,
        available_skills,
    )

    # -----------------------------------------------------
    # Verification
    # -----------------------------------------------------
   
    verified_registry = load_registry(
        registry_path,
    )

    if not any(
        skill.get("id") == new_skill_entry["id"]
        for skill in verified_registry
    ):
        raise RuntimeError(
            "Registry verification failed."
        )

    if not skill_file_path.exists():
        raise RuntimeError(
            "Filesystem verification failed."
        )

    print(
        f"✅ Skill '{new_skill_entry['name']}' "
        "successfully registered."
    )

    return new_skill_entry
````

### Building a planner-readable manifest

````python
def generate_execution_plan(
    user_prompt: str,
    available_skills: List[Dict[str, Any]],
) -> List[Dict[str, Any]]:
    """
    Generate an execution plan for a compound user request.

    Description
    -----------
    Uses the Planner LLM to analyze the user's request and produce
    an ordered workflow consisting of sequential execution steps.

    Each step contains:
        - step_number
        - route
        - task_instruction

    The planner decides whether an existing registered skill can
    perform a step or whether a new skill must first be built.

    Args
    ----
    user_prompt:
        Compound request entered by the user.

    available_skills:
        List of registered skills loaded from the SkillOS registry.

    Returns
    -------
    List[Dict[str, Any]]
        Ordered execution plan.

    Raises
    ------
    RuntimeError
        If the planner returns invalid JSON.

    Example
    -------
    >>> plan = generate_execution_plan(
    ...     "Summarize a report and email it.",
    ...     available_skills,
    ... )
    """

    # --------------------------------------------
    # Build planner-readable skill manifest
    # --------------------------------------------

    skills_manifest = "\n".join(
        [
            (
                f"- id: {skill.get('id')} | "
                f"name: {skill.get('name')} | "
                f"description: {skill.get('description', '')}"
            )
            for skill in available_skills
            if skill.get("enabled", True)
        ]
    )

    if not skills_manifest:
        skills_manifest = "- No skills currently registered."

    # --------------------------------------------
    # Planner prompt
    # --------------------------------------------

    planner_prompt = f"""
You are the SkillOS Workflow Planner.

Your responsibility is to analyze a user's request and convert it into
an ordered execution plan.

=========================
AVAILABLE REGISTERED SKILLS
=========================

{skills_manifest}

=========================
USER REQUEST
=========================

{user_prompt}

=========================
RULES
=========================

1. If the user is asking to create, design, draft, or build exactly ONE new skill,
   return exactly ONE step with route "BUILD".
   Do NOT decompose that skill into sub-steps such as input schema, output schema,
   generation logic, testing, or implementation phases.
   The task_instruction must describe the complete requested skill in one instruction.

2. If the user is asking to create, design, draft, or build MORE THAN ONE new skill,
   return one sequential BUILD step per requested skill.

3. If the user asks for a workflow that uses existing skills and/or new skills,
   break the workflow into chronological operational steps.

4. If an existing registered skill can perform a step,
   set "route" to that skill's exact id.

5. If no existing registered skill can perform the step,
   set "route" to "BUILD".

6. Never invent skill ids.

7. If one step depends on the output of another step,
   explicitly mention that dependency in task_instruction.

8. Preserve chronological order.

9. Output ONLY valid JSON.

10. Do NOT wrap JSON inside Markdown.

=========================
OUTPUT FORMAT
=========================

[
  {{
    "step_number": 1,
    "route": "BUILD",
    "task_instruction": "..."
  }}
]
"""

    print(
        "\n🧠 Planner is analyzing the request "
        "and generating the execution plan..."
    )

    planner_client = genai.Client()
    
    # Pause to respect API rate limits before the next LLM call
    time.sleep(API_REQUEST_DELAY)

    planner_response = planner_client.models.generate_content(
        model=PLANNER_MODEL,
        contents=planner_prompt,
    )

    try:

        execution_plan = extract_json(planner_response.text)
        print("\n📋 Execution Plan\n")
        print(
            json.dumps(
                execution_plan,
                indent=2,
                ensure_ascii=False,
            )
        )

        return execution_plan

    except Exception as error:
        print("\n❌ Planner produced invalid JSON.\n")
        print(planner_response.text)
        raise RuntimeError(
            "Planner failed to generate a valid execution plan."
        ) from error
````

## Build Functions

### Generating a new SKILL.md using the Skill Factory

````python
async def build_skill(
    instruction: str,
    step_number: int,
) -> str:
    """
    Generate a new SKILL.md using the Skill Factory.

    Description
    -----------
    Creates a temporary Skill Factory agent, sends the user's
    instruction, and returns the generated SKILL.md draft.

    This function does NOT:

    - ask for human approval,
    - write files,
    - update the registry,
    - execute the generated skill.

    Those responsibilities belong to later pipeline stages.

    Args
    ----
    instruction:
        The task description that should be converted into a SKILL.md.

    step_number:
        Workflow step number.

    Returns
    -------
    str
        Generated SKILL.md draft.

    Raises
    ------
    RuntimeError
        If no draft is produced.

    Example
    -------
    >>> draft = await build_skill(
    ...     instruction="Summarize PDF documents.",
    ...     step_number=1,
    ... )
    """

    print(
        f"\n🛠️ Activating Skill Factory "
        f"for Step {step_number}..."
    )

    # -----------------------------------------------------
    # Create generator model
    # -----------------------------------------------------

    skill_factory_model = Gemini(
        model=PLANNER_MODEL, # Use the planner model as the skill manager model
        api_key=os.environ["GOOGLE_API_KEY"],
    )

    # -----------------------------------------------------
    # Load master SKILL.md
    # -----------------------------------------------------

    master_skill_object = load_skill_from_dir(
        str(FACTORY_SKILL_PATH)
    )

    master_skill_toolset = SkillToolset(
        skills=[master_skill_object]
    )

    # -----------------------------------------------------
    # Create Skill Factory agent
    # -----------------------------------------------------

    skill_factory_agent = Agent(
        name="skill_factory",
        model=skill_factory_model,
        instruction=(
            "You are the Skill Factory Agent.\n"
            "Your only output must be the complete raw contents of one SKILL.md file.\n"
            "The first characters of your response must be exactly three hyphens: ---\n"
            "Do not write explanations, plans, reasoning, introductions, status messages, or commentary.\n"
            "Do not wrap the output in markdown fences.\n"
            "Do not write ```markdown or ```.\n"
            "Start immediately with YAML frontmatter.\n"
            "The required format is:\n"
            "---\n"
            "name: kebab-case-skill-name\n"
            "description: One concise sentence describing when to use this skill.\n"
            "---\n\n"
            "# Skill Title\n\n"
            "Then write the full skill instructions.\n"
            "Do NOT write files.\n"
            "Do NOT modify registries.\n"
            "Wait for external human approval."
        ),
        tools=[master_skill_toolset],
    )

    # -----------------------------------------------------
    # Create runtime session
    # -----------------------------------------------------

    app_name = f"factory_app_{step_number}"
    user_id = "user_1"
    session_id = str(uuid.uuid4())

    session_service = InMemorySessionService()

    runner = Runner(
        agent=skill_factory_agent,
        app_name=app_name,
        session_service=session_service,
    )

    await session_service.create_session(
        app_name=app_name,
        user_id=user_id,
        session_id=session_id,
    )

    # -----------------------------------------------------
    # Build message
    # -----------------------------------------------------

    message = types.Content(
        role="user",
        parts=[
            types.Part(
                text=instruction,
            )
        ],
    )

    # -----------------------------------------------------
    # Execute Skill Factory
    # -----------------------------------------------------

    draft_chunks: List[str] = []

    # Pause to respect API rate limits before the next LLM call
    await asyncio.sleep(API_REQUEST_DELAY)

    async for event in runner.run_async(
        user_id=user_id,
        session_id=session_id,
        new_message=message,
    ):

        if (
            event.is_final_response()
            and event.content
            and event.content.parts
        ):

            segment = "\n".join(
                part.text
                for part in event.content.parts
                if part.text
            )

            if segment:
                draft_chunks.append(segment)

    if not draft_chunks:
        raise RuntimeError(
            "Skill Factory failed to generate a SKILL.md draft."
        )

    full_draft = "\n\n".join(draft_chunks)

    # Extract the first valid SKILL.md block
    full_draft = clean_generated_skill_md(full_draft)

    print("\n================ GENERATED SKILL DRAFT ================\n")
    print(full_draft)
    print("\n=======================================================\n")

    return full_draft
````

### Judging a newly generated skill

````python
async def judge_skill(
    artifact: str,
    rubric: Dict[str, Any] = SKILL_CREATION_RUBRIC,
    minimum_score: float = 4.0,
    judge_model: str = JUDGE_MODEL,
) -> Dict[str, Any]:
    """
    Evaluate an artifact using a configurable LLM-as-Judge rubric.

    Description
    -----------
    Uses an LLM to evaluate any generated artifact according to a
    supplied rubric configuration. The rubric defines the evaluation
    title and criteria, making this function reusable across SkillOS.

    This function does NOT:

    - regenerate artifacts,
    - request human approval,
    - write files,
    - modify the registry.

    Args
    ----
    artifact:
        Text to evaluate (for example, a generated SKILL.md).

    rubric:
        Rubric configuration dictionary. Expected format:

        {
            "title": "...",
            "criteria": [
                "...",
                "...",
                ...
            ]
        }

    minimum_score:
        Minimum acceptable average score.

    judge_model:
        Model used for evaluation.

    Returns
    -------
    Dict[str, Any]

        {
            "report": {...},
            "scores": {...},
            "average_score": float,
            "passed": bool,
            "summary": str,
        }

    Raises
    ------
    ValueError
        If the rubric is invalid.

    RuntimeError
        If the judge model fails or returns invalid JSON.

    Example
    -------
    >>> result = await judge_skill(
    ...     artifact=full_draft,
    ...     rubric=SKILL_CREATION_RUBRIC,
    ... )
    """

    print("⚖️ LLM-as-Judge is evaluating the draft quality...")
    
    if "title" not in rubric:
        raise ValueError("Rubric must contain a 'title' field.")

    if "criteria" not in rubric:
        raise ValueError("Rubric must contain a 'criteria' field.")

    if not rubric["criteria"]:
        raise ValueError("Rubric criteria cannot be empty.")

    criteria_text = "\n".join(
        f"{i + 1}. {criterion}"
        for i, criterion in enumerate(rubric["criteria"])
    )

    print(f"\n⚖️ Running judge: {rubric['title']}...\n")

    prompt = f"""
You are an expert evaluator.

Evaluate the following artifact using the evaluation criteria below.

Evaluation Title:
{rubric["title"]}

Evaluation Criteria:

{criteria_text}

Assign a score to every criterion using the following scale:

5 = Excellent (fully satisfies the criterion)
4 = Good (minor improvements possible)
3 = Acceptable (adequate but noticeable weaknesses)
2 = Poor (major deficiencies)
1 = Very Poor (fails to satisfy the criterion)

Use only the numbers 1, 2, 3, 4, or 5.

The average_score must be the arithmetic mean of all criterion scores and may contain decimals.

Return ONLY valid JSON.

JSON schema:

{{
  "scores": {{
    "<criterion>": 1
  }},
  "average_score": 0.0,
  "summary": ""
}}

Artifact:

\"\"\"
{artifact}
\"\"\"
"""
    
    judge_client = genai.Client()

    # Pause to respect API rate limits before the next LLM call
    await asyncio.sleep(API_REQUEST_DELAY)
    
    try:
        # ALTERNATIVE - 1:
        response = judge_client.models.generate_content(
            model=judge_model,
            contents=prompt,
        )

         # ALTERNATIVE - 2:
        # response = openai_client.responses.create(
        #     model=judge_model,   
        #     input=prompt,
        # )
   
    except Exception as error:
        raise RuntimeError("Judge model execution failed.") from error

    try:
        # ALTERNATIVE - 1:
        evaluation = extract_json(response.text)

        # ALTERNATIVE - 2:
        # evaluation = extract_json(response.output_text)
    except Exception as error:
        raise RuntimeError("Judge returned invalid JSON.") from error

    average_score = float(
        evaluation.get("average_score", 0.0)
    )

    result = {
        "report": evaluation,
        "scores": evaluation.get("scores", {}),
        "average_score": average_score,
        "passed": average_score >= minimum_score,
        "summary": evaluation.get("summary", ""),
    }

    print("\n================ JUDGE RESULT ================\n")
    print(
        json.dumps(
            result,
            indent=2,
            ensure_ascii=False,
        )
    )
    print("\n==============================================\n")

    return result
````

### Getting human approval

````python
async def get_human_approval(
    step_number: int,
    skill_name: str,
    judge_result: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Request human approval before registering a generated skill.

    Description
    -----------
    Displays the LLM-as-Judge evaluation results and pauses the
    SkillOS workflow until the user either approves the generated
    skill or provides revision feedback.

    This function implements the mandatory Human Approval Gate
    defined in the Master SKILL.md.

    No filesystem operations, registry updates, or skill execution
    should occur before this function returns an approved status.

    Args
    ----
    step_number:
        Workflow step currently being executed.

    skill_name:
        Human-readable name (or ID) of the generated skill.

    judge_result:
        Dictionary returned by `judge_skill()`.

    Returns
    -------
    Dict[str, Any]

        {
            "approved": bool,
            "feedback": str
        }

    Example
    -------
    >>> approval = await get_human_approval(
    ...     step_number=1,
    ...     skill_name="market-analysis",
    ...     judge_result=result,
    ... )

    >>> if approval["approved"]:
    ...     ...

    Notes
    -----
    Human approval is completely separate from the LLM-as-Judge.

    The judge only determines whether the draft passes the quality
    threshold.

    Only the human can approve registration.
    """

    print("\n================ HUMAN APPROVAL GATE ================\n")

    print(f"Workflow Step : {step_number}")
    print(f"Skill         : {skill_name}")

    print(
        f"Judge Score   : "
        f"{judge_result['average_score']:.2f}/5.00"
    )

    print(
        f"Passed Judge  : "
        f"{judge_result['passed']}"
    )

    print("\nJudge Summary\n")

    print(judge_result.get("summary", ""))

    print("\n=====================================================\n")

    feedback = input(
        "Type 'APPROVED' to register this skill,\n"
        "or enter revision feedback:\n\n> "
    ).strip()

    approved = feedback.upper() == "APPROVED"

    return {
        "approved": approved,
        "feedback": feedback,
    }
````

## Workflow Orchestration Functions

### Processing and finalizing a new skill, and producing its output

````python
async def execute_build_step(
    step: Dict[str, Any],
    available_skills: List[Dict[str, Any]],
    contextual_pipeline_data: str,
    user_selection: Optional[str] = None,
) -> str | WorkflowPause:
    """
    Execute a workflow step that builds and registers a new skill.

    Description
    -----------
    Orchestrates the complete SkillOS BUILD pipeline.

    Pipeline:

        1. Generate a SKILL.md.
        2. Evaluate it using the LLM-as-Judge.
        3. Obtain mandatory human approval.
        4. Register the skill.
        5. Return a success message.

    Unlike previous versions of the BUILD pipeline, this function
    intentionally does not execute the newly created skill after
    registration.

    Args
    ----
    step:
        One execution step produced by the planner.

    available_skills:
        Current skill registry loaded in memory.

    contextual_pipeline_data:
        Outputs generated by previous workflow steps.

    user_selection:
        Optional user input supplied when resuming a paused workflow.

    Returns
    -------
    str | WorkflowPause

        Confirmation that the skill was successfully built and registered.

    Raises
    ------
    RuntimeError
        If any stage of the BUILD pipeline fails.

    Example
    -------
    >>> message = await execute_build_step(
    ...     step=step,
    ...     available_skills=available_skills,
    ...     contextual_pipeline_data=context,
    ... )
    """

    # -----------------------------------------------------
    # User selection from a resumed workflow
    # -----------------------------------------------------

    if user_selection is not None:
        print(
            f"Using user selection: {user_selection}"
        )

    step_number = step["step_number"]
    instruction = step["task_instruction"]

    # -----------------------------------------------------
    # Build the prompt for skill generation
    # -----------------------------------------------------

    current_prompt = instruction

    if contextual_pipeline_data:
        current_prompt += (
            "\n\nContext/Data from previous workflow steps:\n"
            f"{contextual_pipeline_data}"
        )

    if user_selection is not None:
        current_prompt += (
            "\n\nUser Selection:\n"
            f"{user_selection}"
        )

    print("▶️ Execution step - 2")
    
    # -----------------------------------------------------
    # Stage 1
    # Generate the new skill
    # -----------------------------------------------------

    draft = await build_skill(
        instruction=current_prompt,
        step_number=step_number,
    )

    # -----------------------------------------------------
    # Stage 2
    # Evaluate the generated skill
    # -----------------------------------------------------

    judge_result = await judge_skill(
        artifact=draft,
        rubric=SKILL_CREATION_RUBRIC,
    )

    # -----------------------------------------------------
    # Stage 3
    # Obtain mandatory human approval
    # -----------------------------------------------------

    approval = await get_human_approval(
        step_number=step_number,
        skill_name="Generated Skill",
        judge_result=judge_result,
    )

    if not approval["approved"]:
        raise RuntimeError(
            "Human approval was not granted.\n\n"
            f"Feedback:\n{approval['feedback']}"
        )

    # -----------------------------------------------------
    # Stage 4
    # Register the approved skill
    # -----------------------------------------------------

    registered_skill = register_skill(
        full_draft=draft,
        available_skills=available_skills,
    )

    # -----------------------------------------------------
    # Stage 5
    # Return success
    # -----------------------------------------------------

    message = (
        f"Skill '{registered_skill['name']}' "
        f"was built and registered successfully "
        f"for Step {step_number}."
    )

    print(f"\n✅ {message}\n")

    return message
````

### Generating an output using an existing skill

````python
async def execute_existing_step(
    step: Dict[str, Any],
    available_skills: List[Dict[str, Any]],
    contextual_pipeline_data: str,
    user_selection: Optional[str] = None,
) -> str | WorkflowPause:
    
    """
    Execute a workflow step using an already registered SkillOS skill.

    Description
    -----------
    Locates the requested skill in the SkillOS registry, prepares the
    prompt (including outputs from previous workflow steps), executes
    the skill, and returns its output.

    Unlike execute_build_step(), this function does NOT:

    - build a new skill,
    - invoke the Skill Factory,
    - evaluate the skill,
    - request human approval,
    - update the registry.

    It simply executes an existing registered skill.

    Args
    ----
    step:
        One execution step produced by the Planner.

    available_skills:
        Registry loaded into memory.

    contextual_pipeline_data:
        Outputs generated by previous workflow steps.

    user_selection:
        Optional user input supplied when resuming a paused workflow.

    Returns
    -------
    str | WorkflowPause

        Output produced by the existing skill.

    Raises
    ------
    RuntimeError
        If the requested skill cannot be found.

    Example
    -------
    >>> output = await execute_existing_step(
    ...     step=step,
    ...     available_skills=available_skills,
    ...     contextual_pipeline_data=context,
    ... )
    """


    # -----------------------------------------------------
    # User selection from a resumed workflow
    # -----------------------------------------------------
    
    if user_selection is not None:
        print(
            f"Using user selection: {user_selection}"
        )
    
    # -----------------------------------------------------
    # Extract planner information
    # -----------------------------------------------------
    
    step_number = step["step_number"]
    route = step["route"]
    instruction = step["task_instruction"]

    print(
        f"\n▶️ Executing existing skill "
        f"'{route}' "
        f"(Step {step_number})...\n"
    )

    # -----------------------------------------------------
    # Find registered skill
    # -----------------------------------------------------
    
    selected_skill = find_skill_by_route(
        route=route,
        available_skills=available_skills,
    )

    if selected_skill is None:
        raise RuntimeError(
            f"No registered skill found for route '{route}'."
        )

    output_schema = OUTPUT_SCHEMAS.get(selected_skill["id"])

    # -----------------------------------------------------
    # Build execution prompt
    # -----------------------------------------------------
    
    current_prompt = instruction
    
    if contextual_pipeline_data:
    
        current_prompt += (
            "\n\nContext/Data from previous workflow steps:\n"
            f"{contextual_pipeline_data}"
        )
    
    if user_selection is not None:
    
        current_prompt += (
            "\n\nUser Selection:\n"
            f"{user_selection}"
        )

    # -----------------------------------------------------
    # Execute skill
    # -----------------------------------------------------

    print("▶️ Execution step - 2")

    output = await run_skill_agent(
        skill_path=Path(selected_skill["skill_path"]),
        prompt=current_prompt,
        step_num=step_number,
        output_schema=output_schema,
    )

    output = execute_runtime_command_if_needed(output)

    print(
        f"\n✅ Existing skill "
        f"'{selected_skill['name']}' "
        f"completed successfully.\n"
    )

    return output
````

### Executing a complete SkillOS workflow

````python
async def run_skillos(
    user_incoming_prompt: str,
    *,
    user_selection: Optional[str] = None,
) -> WorkflowResult | WorkflowPause:
    """
    Execute a complete SkillOS workflow.

    Description
    -----------
    This is the primary runtime entry point of SkillOS.

    The workflow consists of:

        1. Load the registered skills.
        2. Generate an execution plan.
        3. Execute each workflow step.
        4. Accumulate workflow context.
        5. Return either a completed WorkflowResult or a WorkflowPause.

    Args
    ----
    user_incoming_prompt:
        Compound request supplied by the user.

    user_selection:
        Optional user-selected value used when resuming a paused workflow.

        Example:
            If SkillOS asks the user to choose one brand name from several
            alternatives, this parameter receives the selected brand name
            when the workflow is resumed.

    Returns
    -------
    WorkflowResult | WorkflowPause

    WorkflowResult
        Returned when the workflow completes successfully or fails.

    WorkflowPause
        Returned when the workflow needs the user to choose one option
        before execution can continue.

    Example
    -------
    >>> result = await run_skillos(
    ...     "Suggest five brand names for an eco-friendly clothing company."
    ... )

    >>> if isinstance(result, WorkflowResult):
    ...     print(result.workflow_output)

    >>> if isinstance(result, WorkflowPause):
    ...     print(result.options)
    """

       
    print("\n========================================================")
    print("🚀 SkillOS Runtime Started")
    print("========================================================\n")
    # -----------------------------------------------------
    # Load registered skills
    # -----------------------------------------------------

    registered_skills = load_registry(REGISTRY_FILE_PATH)

    print(
        f"📚 Loaded {len(registered_skills)} registered skill(s).\n"
    )

    # -----------------------------------------------------
    # Display incoming request
    # -----------------------------------------------------

    print("📥 User Request\n")
    print(user_incoming_prompt)
    print()

    # -----------------------------------------------------
    # Handle resumed workflow input
    # -----------------------------------------------------

    resumed_context = ""

    if user_selection is not None:
        resumed_context = (
            "\n\n[User Selection]\n"
            f"{user_selection}"
        )

        print("🔄 Resuming workflow with user selection:")
        print(user_selection)
        print()

    # -----------------------------------------------------
    # Generate execution plan
    # -----------------------------------------------------

    execution_plan = generate_execution_plan(
        user_prompt=user_incoming_prompt,
        available_skills=registered_skills,
    )

    print("▶️ Execution step - 1")
    
    # -----------------------------------------------------
    # Workflow context
    # -----------------------------------------------------

    contextual_pipeline_data = resumed_context

    # -----------------------------------------------------
    # Execute planner steps
    # -----------------------------------------------------

    try:

        for step in execution_plan:

            step_number = step["step_number"]
            route = step["route"]
            instruction = step["task_instruction"]

            print("\n--------------------------------------------------------")
            print(f"▶️ STEP {step_number}")
            print("--------------------------------------------------------")
            print(f"Route       : {route}")
            print(f"Instruction : {instruction}\n")

            # -------------------------------------------------
            # Build a new skill
            # -------------------------------------------------

            if route == "BUILD":

                step_output = await execute_build_step(
                    step=step,
                    available_skills=registered_skills,
                    contextual_pipeline_data=contextual_pipeline_data,
                    user_selection=user_selection,
                )

            # -------------------------------------------------
            # Execute an existing skill
            # -------------------------------------------------

            else:

                step_output = await execute_existing_step(
                    step=step,
                    available_skills=registered_skills,
                    contextual_pipeline_data=contextual_pipeline_data,
                    user_selection=user_selection,
                )

            # Pause to respect API rate limits before the next LLM call
            await asyncio.sleep(API_REQUEST_DELAY)
            
            # -------------------------------------------------
            # WorkflowPause
            # -------------------------------------------------

            # Some skills may require the user to choose one
            # option before the workflow can continue.
            #
            # For example:
            #
            # • Choose one brand name
            # • Choose one slogan
            # • Choose one logo
            # • Choose one translation
            #
            # In that case immediately return the WorkflowPause
            # to the notebook UI.

            if isinstance(step_output, WorkflowPause):

                print(
                    f"\n⏸️ Workflow paused during Step "
                    f"{step_number}.\n"
                )

                return step_output

            # -------------------------------------------------
            # Store workflow output
            # -------------------------------------------------

            contextual_pipeline_data += (
                f"\n\n"
                f"[Step {step_number} Output]\n"
                f"{step_output}"
            )

            print(f"\n✅ STEP {step_number} completed.\n")

    # -----------------------------------------------------
    # Runtime error
    # -----------------------------------------------------

    except Exception as error:
    
        print("\n========================================================")
        print("❌ SkillOS Execution Failed")
        print("========================================================\n")
    
        traceback.print_exc()
    
        return WorkflowResult(
            status="FAILED",
            execution_plan=execution_plan,
            workflow_output=contextual_pipeline_data,
            error=str(error),
        )
    
    # -----------------------------------------------------
    # Workflow completed successfully
    # -----------------------------------------------------

    print("\n========================================================")
    print("🏁 SkillOS Workflow Completed Successfully")
    print("========================================================\n")

    return WorkflowResult(
        status="SUCCESS",
        execution_plan=execution_plan,
        workflow_output=contextual_pipeline_data,
        error=None,
    )
````

### Printing tool summary

````python
def print_tool_summary(
    tool_call: dict[str, Any],
    provider: str = "GitHub",
) -> None:
    """
    Print a readable summary of a proposed MCP tool call.

    Args
    ----
    tool_call:
        Dictionary containing:
        - tool: selected MCP tool name
        - arguments: arguments passed to the tool

    provider:
        Name of the MCP provider, such as "GitHub".

    Returns
    -------
    None
    """

    # -----------------------------------------------------
    # Extract tool name and arguments
    # -----------------------------------------------------

    tool = tool_call.get("tool")
    args = tool_call.get("arguments", {})

    # -----------------------------------------------------
    # Header
    # -----------------------------------------------------

    print("\n" + "=" * 70)
    print(f"🔧 Proposed {provider} Action")
    print("=" * 70)

    # -----------------------------------------------------
    # Provider-specific summaries
    # -----------------------------------------------------

    if tool == "create_repository":
        print(f"\nRepository : {args.get('name') or args.get('repo_name')}")
        print(f"Visibility : {args.get('visibility', 'public')}")

    elif tool == "delete_repository":
        print(f"\nRepository : {args.get('name') or args.get('repo_name')}")
        print("\n⚠️ This repository will be permanently deleted.")

    elif tool in {"upload_files", "push_files", "create_or_update_file"}:
        print(f"\nRepository  : {args.get('repository') or args.get('repo_name')}")
        print(f"Local Folder: {args.get('local_dir')}")
        print(f"Branch      : {args.get('branch', 'main')}")

    else:
        print(f"\nTool: {tool}")

        print("\nArguments:")
        for key, value in args.items():
            print(f"  • {key}: {value}")

    # -----------------------------------------------------
    # Footer
    # -----------------------------------------------------

    print("\n" + "=" * 70)
````

### LLM invocation of the GitHub tool

````python
async def ask_llm_to_choose_mcp_tool(
    user_request: str,
    tools_text: str,
) -> dict[str, Any]:
    """
    Ask the LLM to select the most appropriate GitHub MCP tool.

    Description
    -----------
    Provides the available GitHub MCP tools together with the user's
    natural-language request, allowing the LLM to select the single
    best tool and generate the corresponding arguments.

    The model must return exactly one JSON object containing:

        {
            "tool": "...",
            "arguments": { ... }
        }

    Args
    ----
    user_request:
        Natural-language description of the GitHub task.

    tools_text:
        Human-readable list of available GitHub MCP tools and their
        descriptions.

    Returns
    -------
    dict[str, Any]
        Dictionary containing the selected tool name and its arguments.

    Raises
    ------
    ValueError
        If the model does not return a valid tool selection.

    Example
    -------
    >>> tool_call = await ask_llm_to_choose_mcp_tool(
    ...     user_request="Create a new repository named project-catflix.",
    ...     tools_text=tools_text,
    ... )
    """

    # -----------------------------------------------------
    # Build the LLM prompt
    # -----------------------------------------------------

    prompt = f"""
You may call exactly one GitHub MCP tool.

Available GitHub MCP tools:

{tools_text}

User request:

{user_request}

Select the single most appropriate tool and generate its arguments.

Return only a valid JSON object in the following format:

{{
    "tool": "...",
    "arguments": {{
        ...
    }}
}}
"""

    # -----------------------------------------------------
    # Ask the LLM to choose the MCP tool
    # -----------------------------------------------------

    response = openai_client.responses.create(
        model=GITHUB_MODEL,
        input=prompt,
    )

    # -----------------------------------------------------
    # Parse the returned JSON
    # -----------------------------------------------------

    tool_call = extract_json(
        response.output_text,
    )

    # -----------------------------------------------------
    # Validate the returned structure
    # -----------------------------------------------------

    if (
        "tool" not in tool_call
        or "arguments" not in tool_call
    ):
        raise ValueError(
            "The LLM did not return a valid MCP tool selection."
        )

    return tool_call
````

### Getting tool approval

````python
async def get_mcp_tool_approval(
    provider: str,
    tool_call: dict[str, Any],
) -> dict[str, Any]:
    """
    Obtain human approval before invoking an MCP tool.

    Description
    -----------
    Displays a human-readable summary of the selected MCP tool and
    prompts the user to approve or reject the proposed action before
    it is executed.

    This function is intended for operations that modify external
    systems, such as GitHub repositories, cloud resources, or other
    MCP-enabled services.

    Args
    ----
    provider:
        Name of the MCP provider, such as "GitHub".

    tool_call:
        Dictionary describing the selected MCP tool and the arguments
        that will be passed to it.

    Returns
    -------
    dict[str, Any]

        Dictionary containing the user's approval decision.

        Example:

        {
            "approved": True
        }

    Example
    -------
    >>> approval = await get_tool_approval(
    ...     provider="GitHub",
    ...     tool_call=tool_call,
    ... )

    Notes
    -----
    The proposed tool invocation is not executed by this function.
    It only collects the user's approval decision.
    """

    # -----------------------------------------------------
    # Display the proposed MCP action
    # -----------------------------------------------------

    print_tool_summary(
        tool_call=tool_call,
        provider=provider,
    )

    # -----------------------------------------------------
    # Ask the user for approval
    # -----------------------------------------------------

    answer = input(
        "\nApprove this action? (y/n): "
    ).strip().lower()

    # -----------------------------------------------------
    # Return the approval decision
    # -----------------------------------------------------

    return {
        "approved": answer in {
            "y",
            "yes",
        },
    }
````

### Executing a GitHub task

````python
async def run_github_mcp_request(
    user_request: str,
    *,
    debug: bool = False,
) -> Any | None:
    """
    Execute a GitHub operation through the GitHub MCP server.

    Description
    -----------
    Connects to the GitHub MCP server, retrieves available tools,
    asks the LLM to choose the appropriate tool, shows the proposed
    action to the user, obtains approval, executes the selected tool,
    and returns the MCP result.

    Args
    ----
    user_request:
        Natural-language GitHub task.

    debug:
        If True, print all available MCP tools and input schemas.

    Returns
    -------
    Any | None
        MCP tool result if executed successfully, or None if the user
        cancels the action or an error occurs.
    """

    # -----------------------------------------------------
    # Build authentication headers
    # -----------------------------------------------------

    headers = {
        "Authorization": f"Bearer {GITHUB_TOKEN}",
    }

    try:
        # -------------------------------------------------
        # Connect to GitHub MCP server
        # -------------------------------------------------

        async with streamablehttp_client(
            GITHUB_MCP_URL,
            headers=headers,
        ) as (read, write, _):

            async with ClientSession(read, write) as session:
                await session.initialize()

                # -----------------------------------------
                # List available MCP tools
                # -----------------------------------------

                tools = await session.list_tools()

                available_tool_names = {
                    tool.name
                    for tool in tools.tools
                }

                tools_text = "\n\n".join(
                    (
                        f"Tool: {tool.name}\n"
                        f"Description: {tool.description}\n"
                        f"Input schema: {tool.inputSchema}"
                    )
                    for tool in tools.tools
                )

                if debug:
                    print("\nAvailable MCP tools:\n")

                    for tool in tools.tools:
                        print("=" * 80)
                        print(tool.name)
                        print(tool.description)
                        print(tool.inputSchema)

                # -----------------------------------------
                # Let LLM choose the MCP tool
                # -----------------------------------------

                tool_call = await ask_llm_to_choose_mcp_tool(
                    user_request=user_request,
                    tools_text=tools_text,
                )

                # -----------------------------------------
                # Handle unsupported requests
                # -----------------------------------------

                if tool_call.get("tool") is None:
                    print(
                        "\n🚫 No suitable GitHub MCP tool is available.\n"
                    )
                    print(tool_call.get("reason", "No reason provided."))
                    return None

                selected_tool = tool_call["tool"]

                if selected_tool not in available_tool_names:
                    print(
                        f"\n🚫 LLM selected unavailable MCP tool: "
                        f"{selected_tool}\n"
                    )
                    return None
                else:
                    print(
                        f"\n🔧 LLM selected the MCP tool: "
                        f"{selected_tool}\n"
                    )

                # -----------------------------------------
                # Show proposed action and ask approval
                # -----------------------------------------

                approval = await get_mcp_tool_approval(
                    provider="GitHub",
                    tool_call=tool_call,
                )

                if not approval["approved"]:
                    print(
                        "\n🚫 GitHub operation cancelled by the user.\n"
                    )
                    return None

                # -----------------------------------------
                # Execute selected MCP tool
                # -----------------------------------------

                print("\n🚀 Executing GitHub MCP tool...\n")

                result = await session.call_tool(
                    selected_tool,
                    arguments=tool_call["arguments"],
                )

                # -----------------------------------------
                # Display result
                # -----------------------------------------

                print(result)

                if hasattr(result, "content"):
                    print(result.content)

                return result

    except ExceptionGroup as eg:
        # -------------------------------------------------
        # Handle ExceptionGroup errors from MCP/AnyIO
        # -------------------------------------------------

        print("\n❌ GitHub MCP operation failed.\n")

        for exc in eg.exceptions:
            traceback.print_exception(
                type(exc),
                exc,
                exc.__traceback__,
            )

        return None
````

This notebook UI collects the selected skill, optional input data, and task description, lets the user load predefined scenarios, and stores the submitted values in shared state before starting the SkillOS workflow.

````python
# ---------------------------------------------------------
# GLOBAL NOTEBOOK STATE
# ---------------------------------------------------------

state = {
    "selected_item": None,
    "description_for_skill": None,
    "data": None,
    "workflow_running": False,
}


# ---------------------------------------------------------
# HELPER: MENU HTML RENDERING
# ---------------------------------------------------------

def render_menu_items(options: List[str]) -> str:
    """
    Render all available menu items.
    """

    return f"""
    <div style="
        font-weight:bold;
        font-size:1.1em;
        margin-bottom:12px;
    ">
        Available Menu Items:
    </div>

    <div style="line-height:1.8;">
        {"<br>".join(options)}
    </div>
    """


# ---------------------------------------------------------
# USER INTERFACE
# ---------------------------------------------------------

menu_box = widgets.HTML(
    value=render_menu_items(OPTIONS),
    layout=widgets.Layout(
        width="700px",
        border="1px solid #d0d0d0",
        padding="12px",
        margin="0 0 16px 0",
    ),
)

scenario_dropdown = widgets.Dropdown(
    options=["Select a scenario..."] + list(SCENARIOS.keys()),
    value="Select a scenario...",
    description="Scenario:",
    layout=widgets.Layout(
        width="700px",
        margin="0 0 15px 0",   # bottom margin = 15px
    ),
)

skill_menu_item = widgets.Text(
    description="Existing skills:",
    placeholder="e.g. Create a brand name",
)

skill_data = widgets.Text(
    description="Input data:",
    placeholder="e.g. current_data",
)

skill_description = widgets.Textarea(
    description="Description:",
    placeholder=(
        "Briefly describe what the skill should do.\n"
        "Example: Create five brand names and continue "
        "with the selected one."
    ),
    layout=widgets.Layout(
        width="700px",
        height="360px",
    ),
)

submit_button = widgets.Button(
    description="Submit",
    button_style="success",
    icon="check"
)

clear_button = widgets.Button(
    description="Clear",
    button_style="warning",
    icon="trash",
)

output = widgets.Output()


# ---------------------------------------------------------
# CALLBACKS
# ---------------------------------------------------------

def on_scenario_change(change):

    if change["name"] != "value":
        return

    if change["new"] == "Select a scenario...":
        return

    scenario = SCENARIOS[change["new"]]

    skill_menu_item.value = scenario["selected_item"].strip()
    skill_data.value = scenario["data"].strip()
    skill_description.value = scenario["description"].strip()


def on_submit(button) -> None:  

    with output:
    
        output.clear_output()
        
        selected_item = skill_menu_item.value.strip()
        data_key = skill_data.value.strip()
        description = skill_description.value.strip()

        if data_key and data_key in DATA:
            data = DATA[data_key]
            description = (
                f"{description}\n\n"
                f"Data:\n\n"
                f"{json.dumps(data, indent=2)}"
            )

            state["data"] = data
        else:
            state["data"] = None

        state["selected_item"] = selected_item
        state["description_for_skill"] = description

        if selected_item and selected_item not in OPTIONS:
            OPTIONS.append(selected_item)
            menu_box.value = render_menu_items(OPTIONS)

        print("Selected Menu Item\n")
        print("------------------\n")
        print(f"{selected_item}\n\n")
        print("Skill Description\n")
        print("------------------\n")
        print(f"{description}\n\n")
        print("✅ Ready. Click ▶ Run SkillOS to start execution.\n")


def on_clear(button) -> None:
    skill_menu_item.value = ""
    skill_data.value = ""
    skill_description.value = ""
    scenario_dropdown.value = "Select a scenario..."
    state["selected_item"] = None
    state["description_for_skill"] = None
    state["data"] = None
    state["workflow_running"] = False
    output.clear_output()
    run_output.clear_output()


scenario_dropdown.observe(on_scenario_change,names="value")
submit_button.on_click(on_submit)
# run_button.on_click(on_run_button_clicked)  
clear_button.on_click(on_clear)

# ---------------------------------------------------------
# DISPLAY NOTEBOOK UI
# ---------------------------------------------------------

display(
    scenario_dropdown,
    menu_box,
    skill_menu_item,
    skill_data,
    skill_description,
    widgets.HBox([submit_button, clear_button]),
    output,
)
````

## B. Deployment

````python
# -------------------------------------------------------
# Widgets
# -------------------------------------------------------

project_name = widgets.Text(
    value="project-catflix",
    description="Project:",
    placeholder="Enter GitHub repository name",
    layout=widgets.Layout(width="500px"),
)

create_button = widgets.Button(
    description="▣ Create Repository",
    button_style="success",
    layout=widgets.Layout(width="220px", height="40px"),
)

commit_message = widgets.Text(
    value="Add web page",
    description="Commit:",
    placeholder="Enter commit message",
    layout=widgets.Layout(width="550px"),
)

publish_button = widgets.Button(
    description="⮕ Publish Website",
    button_style="success",
    layout=widgets.Layout(width="220px", height="40px"),
)

run_output = widgets.Output()
publish_output = widgets.Output()


def on_click(_):
    
    run_output.clear_output()

    with run_output:
        project = project_name.value.strip()

        if not project:
            print("⚠️ Please enter a project name.")
            return

        create_repo_request = f"""
Create or update a public GitHub repository named {project}.
My GitHub username is {GITHUB_USERNAME}.
"""

        state["create_repo_request"] = create_repo_request

        print(f"✅ Repository request prepared for: {project}")
        print("\nNow run the next cell:")
        print("result = await run_github_mcp_request(state['create_repo_request'])")


def on_publish_clicked(_):
    publish_output.clear_output()

    with publish_output:
        project = project_name.value.strip()

        if not project:
            print("⚠️ Please enter a project name.")
            return

        project_dir = Path("/kaggle/working") / project

        if not project_dir.exists():
            print(f"❌ Project '{project}' not found.")
            return

        print(f"📁 Reading website files from {project_dir / 'web'}")

        files = get_web_files(project_dir)

        message = commit_message.value.strip()
        if not message:
            message = "Add web page"

        publish_web_page_request = f"""
Push all files from /kaggle/working/{project}/web to the repository {GITHUB_USERNAME}/{project}.

owner: {GITHUB_USERNAME}
repo: {project}
branch: main
message: {message}
path: index.html
content: {files}
"""

        state["publish_web_page_request"] = publish_web_page_request

        print(f"✅ Publish request prepared for {GITHUB_USERNAME}/{project}.")
        print("\nNow run the next cell:")
        print("result = await run_github_mcp_request(state['publish_web_page_request'])")


create_button.on_click(on_click)
publish_button.on_click(on_publish_clicked)


# -------------------------------------------------------
# Display
# -------------------------------------------------------

display(
    widgets.VBox([
        project_name,
        commit_message,
        create_button,
        run_output,
        publish_button,
        publish_output,
    ])
)
````

    <li>Click <strong>📦 Create Repository</strong></li>
    <li>Run the following cell to create a new public repository or update an existing one.</li>
    <li>Enter an optional commit message. If left empty, the default message <strong>"Add web page"</strong> is used.</li>
    <li>Click <strong>🚀 Publish Website</strong></li>
    <li>Run the following cell to collect all generated web files from the project's <span class="skillos-highlight-primary">web</span> directory and publish them to the specified GitHub repository through the GitHub MCP workflow.</li>
    <li>The GitHub MCP operation returns the execution result, which is stored in <span class="skillos-highlight-primary">result</span> and displayed in the notebook output.</li>
</ol>

</div>
